```python
import sys
import os

# Add project root to path
sys.path.append(os.path.abspath("../../../"))

from bmad.skills.text_processing import summarize

# Orchestration Logic: Loop with Threshold
texts = [
    "This is short.",
    "This is a slightly longer text that might need summarization.",
    "Short again."
]

results = []
threshold_length = 20

print("Starting Orchestration...")

for i, text in enumerate(texts):
    print(f"Processing item {i+1}: {text}")
    
    # Guardrail: Max length check (Threshold)
    if len(text) > threshold_length:
        print("  -> Above threshold, invoking Skill: summarize")
        summary = summarize.execute(text)
        results.append(summary)
    else:
        print("  -> Below threshold, keeping original")
        results.append(text)

print("Orchestration Complete.")
print("Results:", results)

assert len(results) == 3
assert results[1].startswith("Summary:")
assert results[0] == "This is short."
```
