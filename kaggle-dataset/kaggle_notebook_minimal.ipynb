# Compound Loop — ARC Prize 2026 Submission

Reference implementation of the Compound Loop metacognitive architecture for ARC-AGI-2.

**Paper**: see README in linked repository.
**License**: MIT

```python
import json, sys

# Load solver from attached dataset
sys.path.insert(0, '/kaggle/input/cohezion-arc-solver')

import arc_solver

# Load official evaluation challenges
with open('/kaggle/input/arc-prize-2026/arc-agi_evaluation_challenges.json') as f:
    challenges = json.load(f)

print(f'Tasks: {len(challenges)}')
```

```python
submission = {}

for task_id, task in sorted(challenges.items()):
    try:
        program = arc_solver.search_program(task['train'], max_depth=3, budget=5000)
        if program is not None:
            preds = [arc_solver.apply_program(arc_solver.deepcopy_grid(t['input']), program)
                     for t in task.get('test', [])]
            if preds:
                submission[task_id] = preds
    except Exception:
        pass  # Log and continue

with open('submission.json', 'w') as f:
    json.dump(submission, f)

print(f'Predictions: {len(submission)}/{len(challenges)}')
```
