# BirdCLEF 2026: Bioacoustic Monitoring Challenge - Submission

## Objectives
- Load Google Perch v2 Backbone.
- Load custom Pytorch MLP Head.
- Perform fast CPU-only inference within the 90-minute time limit.

```python
import tensorflow_hub as hub
import torch


# 1. Setup Constants
DATA_ROOT = "/kaggle/input/birdclef-2026"
MODEL_PATH = "/kaggle/input/birdclef-baseline-checkpoints/baseline_epoch_5.pt"
PERCH_V2_URL = (
    "/kaggle/input/bird-vocalization-classifier/tensorflow2/bird-vocalization-classifier/4"
)

print("Environment initialized.")
```

```python
# 2. Load Models
backbone = hub.load(PERCH_V2_URL)
head = torch.nn.Sequential(
    torch.nn.Linear(1536, 512),
    torch.nn.ReLU(),
    torch.nn.Dropout(0.2),
    torch.nn.Linear(512, 256),
    torch.nn.ReLU(),
    torch.nn.Linear(256, 234),
)
head.load_state_dict(torch.load(MODEL_PATH, map_location="cpu"))
head.eval()

print("Models loaded successfully.")
```
